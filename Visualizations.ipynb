```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]
plt.plot(x, y, color='b', marker='o',markerfacecolor='w', markeredgecolor='red',linewidth=2)
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

### Analysis

* The graph shows a **steady increasing trend**, with Y-values rising from **10 to 40** as X increases from 1 to 5.
* The **highest value is 50 at X = 5**, indicating strong overall growth.


```python
plt.figure(figsize=(10, 5))
plt.plot(x, y, linestyle='--', color='red', label='Sales')
plt.legend()
plt.tight_layout()
plt.show()
```

### Analysis

* Sales show a **consistent upward trend**, increasing from **10 to 50** as X increases from 1 to 5.
* The **largest growth occurs between X = 4 and X = 5**, where sales increase from 30 to 40.


```python
import seaborn as sns
import matplotlib.pyplot as plt
flights = sns.load_dataset('flights')
sns.lineplot(data=flights, x='year', y='passengers')
plt.title('Passengers over Time')
plt.show()
```

### Analysis

* The graph shows a **clear upward trend in passenger numbers**, increasing from around **125 in 1949 to about 475 in 1960**.
* There is a **slight slowdown around 1957–1958**, followed by strong growth toward 1960; the shaded area represents the **range/uncertainty** around the trend.


```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
sns.boxplot(data=tips, x='day', y='total_bill', palette='Set2')
plt.title('Bill Amount by Day')
plt.show()
```

### Analysis
* **Sunday has the highest median bill amount**, while Thursday has the lowest median.
* **Saturday and Sunday show greater variability** in bill amounts, with several high-value bills appearing as outliers.

```python
import plotly.express as px
df = px.data.iris()
fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species')
fig.show()
```

Insights from the Iris Scatter Plot
Positive relationship: sepal_width and sepal_length show a slight positive relationship — flowers with larger sepal widths tend to have somewhat larger sepal lengths.
Species separation: The three species — setosa, versicolor, and virginica — form different clusters in the graph.
Setosa: Setosa points are relatively well separated from the other two species, especially because of their shorter sepal lengths.
Versicolor: Versicolor generally occupies the middle range of sepal length and width.
Virginica: Virginica tends to have larger sepal lengths and is generally positioned toward the upper part of the graph.
Overlap: Versicolor and virginica have some overlapping points, meaning sepal measurements alone cannot perfectly distinguish these two species.
Outliers: A few points appear farther away from their species' main cluster, indicating unusual measurements.
Conclusion: Sepal length and sepal width are useful for identifying differences between iris species, but they are not sufficient to completely separate all species.

```python
import plotly.express as px
df = px.data.gapminder().query("country == 'India'")
fig = px.line(df, x='year', y='lifeExp', title='Life Expectancy in India')
fig.update_traces(line=dict(color='green', width=3))
fig.show()
```

Insights from the "Life Expectancy in India" Graph
Overall upward trend: Life expectancy in India shows a clear increase over the years.
Long-term improvement: The graph indicates that people in India have generally been living longer over time.
Steady growth: Life expectancy rises progressively rather than showing large sudden fluctuations.
Historical development: The increasing trend can be associated with improvements in healthcare, sanitation, nutrition, and living conditions over the period represented in the dataset.
Highest life expectancy: The latest year in the dataset has the highest life expectancy compared with earlier years.
Lowest life expectancy: The earliest year in the dataset has the lowest life expectancy.
Business/data insight: The graph demonstrates how a line chart can effectively show changes in a variable over time.
Conclusion: Overall, the data shows a positive long-term trend in India's life expectancy.

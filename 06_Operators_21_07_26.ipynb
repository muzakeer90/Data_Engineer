##Arithmetic Operators

1. Write a program to perform addition, subtraction, multiplication, division, floor division, modulus, and exponentiation on two numbers.

```python
a=12
b=10
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

2. Swap two numbers without using a third variable.

```python
a=10
b=20
a,b=b,a
print("after swapping")
print("a=",a)
print("b=",b)
```

3. Find whether a number is even or odd using the modulus operator.

```python
x=int(input("enter a number:"))
if x%2==0:
  print("even")
else:
  print("odd")
```

4. Reverse a three-digit number using arithmetic operators.

```python
n=1234
rev=0
while n>0:
  rem=n%10
  rev=rev*10+rem
  n=n//10
print(rev)
```

5. Find the last digit and first digit of a given number.

```python
n=35464
last=n%10
first=int(str(n)[0])
print("last digit=",last)
print("first digit=",first)
```

## Membership Operators

25. Check whether a character exists in a string.


```python
x="data engineering"
print("e" in x)
print("z" in x)
print("data" in x)
```

26. Check whether a number exists in a list.


```python
a=[1,2,3,4,"data"]
print(3 in a)
print(5 in a)
```

27. Count how many vowels in a string belong to a predefined vowel list using `in`.

```python
vowels=["a","e","i","o","u"]
x="data engineering"
count=0
for i in x:
  if i in vowels:
    count+=1
print(count)
```

## Mixed Operator Programs

28. Build a simple calculator using arithmetic and comparison operators.

```python
a=int(input("enter a number:"))
b=int(input("enter a number:"))
operation=input("enter operation:")
if operation=="+":
  print(a+b)
elif operation=="-":
  print(a-b)
elif operation=="*":
  print(a*b)
elif operation=="/":
  print(a/b)
```

## Assignment 32: Function Returning a Dictionary

- Return a dictionary containing:
  - Total characters
  - Total words
  - Total vowels
  - Total digits
  - Total spaces

---



```python
def analyze_text(text):
# Analyze a string and return a dictionary of statistics.
    return {
        "Total characters": len(text),
        "Total words": len(text.split()),
        "Total vowels": sum(1 for c in text if c.lower() in "aeiou"),
        "Total digits": sum(1 for c in text if c.isdigit()),
        "Total spaces": text.count(" ")
    }
text = input("Enter text: ")
result = analyze_text(text)
for k, v in result.items():
    print(f"{k}: {v}")
```

- Return `True` when the number is prime.
- Return `False` otherwise.
- Use the function to print all prime numbers from 1 to 100.

### Hint

A prime number has exactly two factors.

---


```python
def is_prime(n):
#Check whether a number is prime.
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True
#use function to print all prime number
for num in range(1, 101):
    if is_prime(num):
        print(num, end=" ")
```

## Assignment 35: Fibonacci Function

- Return the complete Fibonacci series as a list.
- Handle zero and negative term counts.
- Do not use recursion in this assignment.

### Example Input

    Enter number of terms: 7

### Expected Output

    [0, 1, 1, 2, 3, 5, 8]

---

```python
def generate_fibonacci(n):
  if n <= 0:
    return []
  fib_series = [0, 1]
  for i in range(len(fib_series), n):
    fib_series.append(fib_series[-1] + fib_series[-2])
  return fib_series
print(generate_fibonacci(7))
```

Write a function named `create_bank_account()`.

### Hint

Return multiple functions:

    return deposit, withdraw, check_balance

---

```python
def create_bank_account(initial_balance=0):
    balance = initial_balance

    def deposit(amount):
        nonlocal balance
        balance += amount
        return balance

    def withdraw(amount):
        nonlocal balance
        if amount > balance:
            print("Insufficient balance!")
            return balance
        balance -= amount
        return balance

    def check_balance():
        return balance

    return deposit, withdraw, check_balance

deposit, withdraw, check_balance = create_bank_account(1000)

print("Balance after deposit :", deposit(500))
print("Attempted overdraw    :", withdraw(5000))
print("Balance after withdraw:", withdraw(300))
print("Final balance         :", check_balance())
```

## Assignment 93: Function Decorator

Create a decorator named `execution_logger()`.

### Requirements

- The decorator should print a message before the original function executes.
- It should print another message after execution.
- Apply it to an addition function.
- Preserve the original function's return value.

### Expected Pattern

    Starting function...
    Result: 30
    Function completed.

---

```python
def execution_logger(func):
    def wrapper(*args, **kwargs):
        print("Starting function...")
        result = func(*args, **kwargs)
        print("Result:", result)
        print("Function completed.")
        return result
    return wrapper

@execution_logger
def add(a, b):
    return a + b

add(10, 20)
```

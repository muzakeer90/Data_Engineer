### Q31. Parse a receipt line into named groups for banking case 1; use sample reference `ACC-1031` and explain the result.

**Hints**

1. Write representative valid and invalid samples before writing the pattern.
2. Use raw strings, named groups for extraction, and `fullmatch` for whole-field validation.
3. Never treat regex alone as authorization or full semantic validation; verify dates and checksums separately.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Q31 solution
import re
line = "TXN-1001 | ACC-1001 | ₹12,500 | 04/09/2026"
pattern = re.compile(
    r"(?P<txn>TXN-\d{4})\s*\|\s*"
    r"(?P<account>ACC-\d{4})\s*\|\s*"
    r"(?P<amount>₹[\d,]+)\s*\|\s*"
    r"(?P<date>\d{2}/\d{2}/\d{4})"
)
m = pattern.fullmatch(line)
print(m.groupdict() if m else "Invalid receipt")
```

```python
# Q34 solution
import re
phones = ["+91-9876543210", "091 9876543210", "9876543210"]
for phone in phones:
    digits = re.sub(r"\D", "", phone)
    if digits.startswith("91") and len(digits) == 12:
        digits = digits[2:]
    elif digits.startswith("091") and len(digits) == 13:
        digits = digits[3:]
    print(phone, "+91" + digits if len(digits) == 10 else "Invalid")
```
